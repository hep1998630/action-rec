The steps to run the real time action recognition system: 
- Use yolov8 pose estimation to extract keypoints. 
- Model the keypoints in a way that can be understood by ML models. 
- Train an RNN to model the sequence. 
- Write the inference pipeline of the full system. 


```python

import pandas as pd 
import numpy as np 
import cv2 
```

```python
base_dir = "ucf101-action-recognition"
train_data = pd.read_csv(f"{base_dir}/train.csv")
```

```python
train_data 
```

```python
train_data_list = train_data.clip_path.to_list()
video_length_list= []
for video_path in train_data_list: 
    cap = cv2.VideoCapture(f"{base_dir}{video_path}")
    length = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))
    print(length)
    video_length_list.append(length)


```
